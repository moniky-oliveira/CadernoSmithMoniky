<h1 id="historia">História dos Números de Smith</h1>

Na Matemática, muitas descobertas começaram justamente com uma simples observação do cotidiano. Um exemplo disso ocorreu em 1982, quando o matemático canadense-americano Albert Wilansky observou o número de telefone de seu cunhado, Harold Smith: 493-7775. Ele notou que ao fatorar este número, a soma dos algarismos do número se igualava a soma dos algarismos de seus fatores primos.

<figure style="text-align:center;">
    <img src="C:\Users\Jean\Desktop\jupyter book/Apresenta1.jpg" width="700">
</figure>

Wilansky cunhou o termo "número de Smith" em homenagem a Harold Smith. Essa observação deu origem ao estudo dessa classe especial de números.

Abaixo temos uma tabela com os cinco primeiros números com essa mesma propriedade.

| Número | Soma dos algarismos | Fatoração | Soma dos algarismos da fatoração |
|:------:|:-------------------:|:---------:|:-------------------:|
| 4      | 4                   | 2 x 2     | 4                   |
| 22     | 4                   | 2 x 11    | 4                   |
| 27     | 9                   | 3 x 3 x 3 | 9                   |
| 58     | 13                  | 2 x 29    | 13                  |
| 85     | 13                  | 5 x 17    | 13                  |
